```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
features=['sepal_length','sepal_width','petal_length','petal_width']
X_train=data[features]

y_train=data['class']
```

```python
from sklearn.linear_model import LogisticRegression
model_1=LogisticRegression(penalty=None,
                          multi_class='multinomial',
                          max_iter=100,
                          solver='lbfgs')

model_1.fit(X_train,y_train)
print(model_1.coef_)
print(model_1.intercept_)
```

```python
features_test=['sepal_length','sepal_width','petal_length','petal_width']
X_test=data_test[features_test]

y_test=data_test['class']
```

```python
from sklearn.neighbors import KNeighborsClassifier
model_2=KNeighborsClassifier(n_neighbors=20,
                            weights='uniform',
                            algorithm='auto',
                            metric='minkowski')
model_2.fit(X_train,y_train)
```

```python
from sklearn.model_selection import cross_val_score
params_grid=[1, 5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
cv_scores=np.zeros(shape=len(params_grid))
for i in range(len(params_grid)):
    knn=KNeighborsClassifier(n_neighbors = params_grid[i],
                               weights = 'uniform',
                               algorithm = 'auto',
                               metric = 'minkowski')
    scores=cross_val_score(estimator=knn,
                          X=X_train,
                          y=y_train,
                          scoring='accuracy',
                          cv=5,n_jobs=-1)
    cv_scores[i]=scores.mean()
```

```python
params_grid[np.argmax(cv_scores)]
```

```python
import itertools
n_estimator=[50,100,200]
max_features=['sqrt', 'log2', None]
max_depth=[1,3,5]
min_samples_leaf=[5,10,15]
param_grid=list(itertools.product(n_estimator,max_features,max_depth,min_samples_leaf))

param_grid=pd.DataFrame(data=param_grid,
                       index=range(1,82),columns=['n_estimator','max_features','max_depth','min_samples_leaf'])

param_grid
```

```python
from sklearn.model_selection import cross_val_score
cv_scores=np.zeros(shape=len(param_grid))

for i in range(len(param_grid)):
    rf_c=RandomForestClassifier(n_estimators=param_grid.iloc[i,0],
                               max_features=param_grid.iloc[i,1],
                               max_depth=param_grid.iloc[i,2],
                               min_samples_leaf=param_grid.iloc[i,3],
                               ccp_alpha=0,

                               )
    scores=cross_val_score(estimator=rf_c,
                          X=X_train,
                          y=y_train,
                          scoring='accuracy',
                          cv=5,n_jobs=-1)
    cv_scores[i]=scores.mean()
```

```python
param_grid.iloc[np.argmax(cv_scores)]
```

```python
model_5=RandomForestClassifier(n_estimators=param_grid.iloc[np.argmax(cv_scores),0],
                               max_features=param_grid.iloc[np.argmax(cv_scores),1],
                               max_depth=param_grid.iloc[np.argmax(cv_scores),2],
                               min_samples_leaf=param_grid.iloc[np.argmax(cv_scores),3],
                               ccp_alpha=0)

model_5.fit(X_train,y_train)
```

```python
y_test_pred_5=model_5.predict(X_test)
print("Accuracy: ",round(accuracy_score(y_test,y_test_pred_5),2)*100)
```

```python
degree_grid=[2,3,4]
c_reg= 10 ** np.linspace(-3, 3, 7)
params_grid=list(itertools.product(degree_grid,c_reg))
params_grid=pd.DataFrame(data=params_grid,index=range(1,22),columns=['degree_grid','c_reg'])
params_grid
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
from sklearn.svm import SVC
cv_scores=np.zeros(shape=len(params_grid))

for i in range(len(params_grid)):
    svc=SVC(kernel='poly',
           degree=params_grid.iloc[i,0],
           C=params_grid.iloc[i,1])
    scores=cross_val_score(estimator=svc,
                          X=X_train_scaled,
                          y=y_train,
                          cv=5,scoring='accuracy')
    cv_scores[i]=scores.mean()
```

```python
params_grid.iloc[np.argmax(cv_scores)]
```

```python
svc=SVC(kernel='poly',
           degree=params_grid.iloc[np.argmax(cv_scores),0],
           C=params_grid.iloc[np.argmax(cv_scores),1])

model_6=svc.fit(X_train_scaled,y_train)
```

```python
y_test_pred_6=model_6.predict(X_test_scaled)
print("Accuracy: ", round(accuracy_score(y_test,y_test_pred_6),2)*100)
```
